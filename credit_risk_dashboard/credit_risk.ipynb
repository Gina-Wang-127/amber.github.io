# US Corporate Credit Risk Dashboard

### A personal portfolio writeup

This project is a Streamlit-based credit market intelligence dashboard that monitors US corporate credit risk across investment-grade (IG) and high-yield (HY) markets. I built it to turn raw financial time series into a decision-oriented tool that a student, junior credit analyst, or portfolio team could use to quickly understand whether current credit conditions are normal, rich, or stressed.

## Why I Built This

The original idea for this project was much less focused. At first, I was mostly thinking about whether the dataset would be easy to use. That was a useful starting point, but it was not a strong project goal. A good analytics product should not exist just because the data is convenient. It should answer a real question for a real user.

I redesigned the project around a clearer objective: **help users read the US corporate credit market like a credit analyst.**

The guiding question became:

> Are current corporate credit conditions normal, rich, or stressed, and what evidence supports that view?

That question shaped the data pipeline, the calculations, the page structure, and the final user experience.

## Target Users

The dashboard is designed for two related audiences:

- **Students learning credit analysis** who need a structured way to connect spreads, ratings, default risk, and market regimes.
- **Junior credit analysts** who need a fast market briefing before deeper issuer-level research.

The app is intentionally not just a chart gallery. Each page is organized around a decision question, such as whether spreads are historically wide, whether HY is under pressure, or whether current spread compensation matches default risk.

## Technical Architecture

The project is organized to separate data engineering, analytics logic, and presentation:

- `data_load.py` handles ingestion from FRED and static CSV files into BigQuery.
- `data_utils.py` contains data access functions and analytical calculations.
- `app.py` is the Streamlit entry point and executive dashboard.
- `pages/` contains the deeper analytical modules.
- `utils/style.py` centralizes shared CSS and sidebar navigation.
- `utils/perf.py` provides page load timing and optional profiling.
- `test_data_utils.py` tests core calculation and unit-cleaning logic.

This structure keeps the project maintainable: page files focus on user-facing analysis, while `data_utils.py` owns the reusable business logic.

## Data And Metrics

The dashboard uses a mix of live market data and static reference data:

- **FRED / ICE BofA data:** IG OAS, HY OAS, effective yields, and rating-level OAS.
- **Moody's Aaa/Baa yields:** used to calculate the Baa-Aaa spread as a quality-dispersion signal.
- **SIFMA data:** fixed-income market structure and corporate issuance.
- **Default-rate data:** speculative-grade and investment-grade default rates, plus cumulative default rates by rating.

One important detail was unit consistency. FRED OAS values can appear as percentage points, where `0.81` means `81 bp`. I added a cleaning function, `clean_oas_to_basis_points()`, so the app consistently reports OAS in basis points across all pages.

## Analytical Logic

The app turns raw time series into interpretable signals through several steps:

- `compute_spread_zscore()` measures whether current spreads are high or low relative to the rolling one-year history.
- `compute_spread_percentile()` compares today's spread level with the trailing ten-year distribution.
- `regime_flag()` converts a percentile into an easy-to-read market regime label.
- `generate_outlook()` combines spread valuation, IG/HY divergence, Baa-Aaa trend, and default-rate movement into a rule-based strategy signal.

This logic supports the product goal because it does not stop at visualization. It helps users move from "what happened" to "what does it mean?"

## Performance Improvements

I made several optimizations to keep the app responsive and easier to maintain:

- **SQL pushdown:** BigQuery filters rows, selects specific series, computes simple totals, and retrieves latest values before data reaches pandas.
- **Streamlit caching:** `@st.cache_data` reduces repeated BigQuery calls when users switch pages.
- **Incremental ingestion:** FRED time series append only new observations instead of reloading full history every run.
- **Reusable data functions:** common queries and calculations are centralized, reducing duplicated page logic.
- **Load-time monitoring:** `display_load_time()` makes performance visible during development.

These improvements create faster runtime, cleaner code structure, and a better user experience.

## What I Learned

The biggest lesson was that a project needs a clear goal before the technical work can be meaningful. At the beginning, I focused too much on whether the data was easy. Later, I learned to ask better product questions:

- Who is the target user?
- What decision are they trying to make?
- What does the dashboard help them understand faster?
- What should be calculated, cleaned, or simplified before the user sees it?

The first version was not perfect, but the process of refining the goal, polishing the UI, improving the data layer, and adding clearer analytical logic made the final project much stronger.
